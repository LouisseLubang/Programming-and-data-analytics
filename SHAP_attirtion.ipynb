```python
def map_degree_level(degree_string):
    if pd.isna(degree_string):
        return None

    degree_str = str(degree_string)

    # Check for Doctorate
    if any(term in degree_str for term in ["PhD", "Ph.D", "Doctorate", "Doctoral"]):
        return "Doctorate"

    # Check for Masters
    elif any(term in degree_str for term in ["Master", "Masters", "MBA", "MS", "MA", "M.S", "M.A"]):
        return "Masters"

    # Check for Bachelors
    elif any(term in degree_str for term in ["Bachelor", "Bachelors", "BS", "BA", "B.S", "B.A", "BSc", "B.Sc"]):
        return "Bachelors"

    # Check for Associates
    elif any(term in degree_str for term in ["Associate", "Associates", "AS", "AA", "A.S", "A.A"]):
        return "Associates"

    # Check for Major (usually implies Bachelors)
    elif "Major in" in degree_str or "major in" in degree_str:
        return "Bachelors"

    else:
        return "Other"

# Replace the original column (instead of creating a new one)
df2["Highest Degree Obtained"] = df2["Highest Degree Obtained"].apply(map_degree_level)

# Check the results
print(df2["Highest Degree Obtained"].value_counts())
```

Data Preperation




```python
from sklearn.preprocessing import LabelEncoder, StandardScaler
from imblearn.combine import SMOTETomek
```

```python
#Train Test Split
from sklearn.model_selection import train_test_split
```

```python
# Filter for students who have LEFT the program (not on-going)
df0_clean = df0[df0['Program Status'] == 'Concluded'].copy()

# Keep only students with known withdrawal/LOA status
df0_clean = df0_clean.dropna(subset=['Student Status'])

print(f"Original shape: {df0.shape}")
print(f"After removing null targets: {df0_clean.shape}")
```

```python
features = ['Program','Age upon date of entry','Final Grade', 'Gender by Birth', 'Civil Status']
X = df0_clean[features].copy()
y = df0_clean['Student Status']
```

```python
# 3. Handle missing values
for col in X.select_dtypes(include=[np.number]).columns:
    X[col] = X[col].fillna(X[col].median())

for col in X.select_dtypes(include=['object', 'category']).columns:
    X[col] = X[col].fillna(X[col].mode()[0] if len(X[col].mode()) > 0 else 'Unknown')

# 4. Encode categorical variables
le_dict = {}
for col in X.select_dtypes(include=['object', 'category']).columns:
    le = LabelEncoder()
    X[col] = le.fit_transform(X[col].astype(str))
    le_dict[col] = le

# Encode target
le_y = LabelEncoder()
y_encoded = le_y.fit_transform(y)

# 5. Split FIRST, then resample
X_train, X_test, y_train, y_test = train_test_split(
    X, y_encoded, test_size=0.20, random_state=42, stratify=y_encoded
)
```

```python
print(f"\nBefore SMOTE:")
print(f"Training set size: {X_train.shape}")
print(f"Training target distribution: {np.bincount(y_train)}")

# 6. Apply SMOTE only to training data
smt = SMOTETomek(sampling_strategy='auto', random_state=42)
X_train_resampled, y_train_resampled = smt.fit_resample(X_train, y_train)

print(f"\nAfter SMOTE:")
print(f"Training set size: {X_train_resampled.shape}")
print(f"Training target distribution: {np.bincount(y_train_resampled)}")
```

```python
X_train, X_test, y_train, y_test = train_test_split(X_train_resampled, y_train_resampled , test_size=0.25, random_state=35)
```

```python
X_train.shape, X_test.shape, y_train.shape, y_test.shape
```

Random Forest

```python
#RandomForest Algorithm
from sklearn.ensemble import RandomForestClassifier

#Classification Metrics
from sklearn.metrics import confusion_matrix, classification_report, accuracy_score
from sklearn.metrics import precision_score, recall_score, f1_score
```

```python
rf = RandomForestClassifier(
    n_estimators=500,
    max_depth=2,
    min_samples_split=10,
    min_samples_leaf=8,
    max_features='sqrt',
    class_weight='balanced',
    random_state=42,
)
#Fit to the train dataset
rf.fit(X_train_resampled, y_train_resampled)
```

```python
#Get predictions
predict_test = rf.predict(X_test)
```

```python
#Confusion Matrix

#Get Classes
classes = rf.classes_

#Get Confusion Matrix, add the classes as the labels
cm = confusion_matrix(y_test, predict_test, labels=classes)

#Put in DataFrame for easy viewing
cm_df = pd.DataFrame(cm, index=classes, columns=classes)

cm_df
```

```python
print(f"Forest Accuracy: {accuracy_score(y_test, predict_test):.2%}")
print(f"Forest Recall: {recall_score(y_test, predict_test, average='weighted'):.2%}")
print(f"Forest Percision: {precision_score(y_test, predict_test, average='weighted'):.2%}")
print(f"Forest F1: {f1_score(y_test, predict_test, average='weighted'):.2%}")
```
